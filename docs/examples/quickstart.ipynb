# DolphinDB Notebook 快速上手

在 Python 内核环境安装 `dolphindb-extension[notebook]`，在工具栏选择你自己的测试连接。前两个单元格只使用 Python；DDB 示例只创建当前会话变量和内存表。每个代码块是独立单元格。

```python
import numpy as np
import pandas as pd
from dolphindb_extension import ddb_show

frame = pd.DataFrame({'value': [20, 100, 10]}, index=[2, 10, 1])
frame  # pandas 原生输出
```

```python
ddb_show(frame)  # 显式自定义展示；点击索引列排序得到 1、2、10
```

```python
answer = %ddb 1 + 1
assert answer == 2
answer  # Python 原生输出
```

```python
%%ddb
prices = table(1..3 as id, 10.5 11.2 12.3 as price)
print("DolphinDB ready")
select * from prices
```

```python
captured = %ddb select * from prices
captured  # pandas 原生输出
```

```python
%%ddb -o filtered
select * from prices where price > 11
```

```python
filtered  # -o 只赋值；显示 Python 对象仍使用 pandas
```

```python
%ddb_show filtered
```

```python
print('下一行直接来自 DDB，因此自动使用自定义组件')
%ddb sum(1..5)
```

```python
ddb_show({'frame': frame, 'vector': np.arange(12), 'tensor': np.arange(24).reshape(2, 3, 4)})
```

## 关闭 DDB 会话

执行下一格后，DDB 中的普通变量释放。Python 中的 `captured`、`filtered` 等对象仍存在；下一次 DDB 执行会创建新会话。

```python
%ddb_close
assert len(captured) == 3
captured
```
